```python
def audit_data(dataframe):
    """
    Hàm tự động thống kê giá trị khuyết thiếu và giá trị duy nhất cho từng cột
    """
    audit_df = pd.DataFrame({
        "Kiểu dữ liệu": dataframe.dtypes,
        "Số lượng khuyết": dataframe.isna().sum(),
        "Tỷ lệ khuyết (%)": (dataframe.isna().mean() * 100).round(2),
        "Số giá trị duy nhất (Unique)": dataframe.nunique(),
        "Mẫu dữ liệu": dataframe.iloc[0]
    })
    return audit_df.sort_values("Tỷ lệ khuyết (%)", ascending=False)
audit_data(clean_df)
```
