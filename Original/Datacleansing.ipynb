```python
def audit(df):

    """One-glance quality report for any DataFrame."""

    return pd.DataFrame({
        "dtype": df.dtypes,
        "n_missing": df.isna().sum(),
        "pct_missing": (df.isna().mean() * 100).round(1),
        "n_unique": df.nunique(),
        "sample": df.iloc[0]
    })

audit(df).sort_values("pct_missing", ascending=False)
```

```python
def parse_number(value):
    value = value.strip()

    if "," in value and "." not in value:
        parts = value.split(",")

        # 2,5 → decimal
        if len(parts) == 2 and len(parts[1]) <= 2:
            return float(value.replace(",", "."))

        # 1,500 → thousands
        return float(value.replace(",", ""))

    return float(value.replace(",", ""))
```

```python
def get_salary_min(row):
    numbers = row["salary_vnd"]
    salary_type = row["salary_type"]

    if salary_type in ["RANGE", "FROM", "ABOVE"]:
        return numbers[0]

    return None


def get_salary_max(row):
    numbers = row["salary_vnd"]
    salary_type = row["salary_type"]

    if salary_type == "RANGE":
        return numbers[1]

    elif salary_type == "UP_TO":
        return numbers[0]

    return None


clean_df["salary_min_vnd"] = clean_df.apply(
    get_salary_min,
    axis=1
)

clean_df["salary_max_vnd"] = clean_df.apply(
    get_salary_max,
    axis=1
)
```
